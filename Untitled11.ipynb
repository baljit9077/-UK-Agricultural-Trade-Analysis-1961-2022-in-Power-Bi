```python
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data_frame = pd.read_csv('/FAOSTAT_data_en_7-28-2024 (2).csv')
```

```python
data_frame
```

```python
df_pivot = data_frame.pivot_table(index=['Item', 'Year'], columns=['Element', 'Unit'], values='Value', fill_value=0).reset_index()
```

```python

df_pivot
```

```python
df_pivot.columns = [' '.join(col).strip() for col in df_pivot.columns.values]
```

```python
df_pivot.rename(columns={
    'Year': 'Year',
    'Item': 'Item',
    'Export Quantity t': 'Export Quantity (t)',
    'Export Value 1000 USD': 'Export Value (1000 USD)',
    'Import Quantity t': 'Import Quantity (t)',
    'Import Value 1000 USD': 'Import Value (1000 USD)'
}, inplace=True)
```

```python
formatted_df = pd.DataFrame()
```

```python
for item in df_pivot['Item'].unique():
    item_data = df_pivot[df_pivot['Item'] == item]
    item_data['Item'] = item_data['Item'].where(item_data['Year'] == item_data['Year'].min(), '')
    formatted_df = pd.concat([formatted_df, item_data], ignore_index=True)
```

```python
df_pivot['Item'] = df_pivot['Item'].where(df_pivot['Year'] == df_pivot['Year'].min(), '')
```

```python
df_pivot.to_csv('/FAOSTAT_data_en_7-28-2024 (2).csv', index=False)
```
